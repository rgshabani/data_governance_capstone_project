```python
# Read all parts from the landing volume as one DataFrame
df = spark.read \
    .format("csv") \
    .option("header", "true") \
    .option("inferSchema", "false") \
    .load("/Volumes/capstone_project_dev/landing/source_metada")

# Verify
print(f"Rows    : {df.count():,}")
print(f"Columns : {len(df.columns)}")
df.printSchema()
```

```python
# Write as Delta table in bronze
# Bronze should be RAW — no type casting here, silver handles that
df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("capstone_project_dev.bronze.raw_metadata")

print("✅ Bronze Delta table created!")
```

```python
# Step 4 — Verify
verify = spark.table("capstone_project_dev.bronze.raw_metadata")
print(f"Rows in bronze : {verify.count():,}")
verify.display()
```
