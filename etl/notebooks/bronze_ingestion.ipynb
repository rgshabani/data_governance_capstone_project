```python
import dlt
from pyspark.sql import functions as F

@dlt.table(
    name="raw_metadata",
    comment="Bronze layer — raw metadata ingested from landing zone, minimal transformation"
)
@dlt.expect_or_drop("valid_column_id", "column_id IS NOT NULL")
@dlt.expect_or_drop("valid_table_id", "table_id IS NOT NULL")
def raw_metadata():
    return (
        spark.read
        .format("csv")
        .option("header", "true")
        .option("inferSchema", "false")
        .load("/Volumes/capstone_project_dev/landing/source_metada")
        .withColumn("_ingest_timestamp", F.current_timestamp())
        .withColumn("_source_file", F.col("_metadata.file_path"))
    )
```
